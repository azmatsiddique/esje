```python
# Quick one-liner query:
%sql SELECT * FROM sales_data LIMIT 5

# Capture query result directly into a Python DataFrame variable:
df = %sql SELECT country, SUM(revenue) AS total_revenue FROM sales_data GROUP BY country
```

```python
%%sql -o sales_summary
SELECT 
    category,
    COUNT(*) AS total_orders,
    SUM(revenue) AS total_revenue
FROM sales_data
GROUP BY category
ORDER BY total_revenue DESC;
```

```python
%%sql --live
SELECT category, SUM(revenue) AS total_revenue 
FROM sales_data 
GROUP BY category;

import matplotlib.pyplot as plt

df.plot(
    x='category', 
    y='total_revenue', 
    kind='bar', 
    title='Total Revenue by Category',
    color='skyblue',
    figsize=(8, 4)
)
plt.ylabel('Revenue ($)')
plt.tight_layout()
plt.show()
```
